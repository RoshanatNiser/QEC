```python
"""
Shor's 9-Qubit Code — Combined Depolarizing + Global Crosstalk Noise
================================================================
Eq.1 : single-qubit depolarizing noise applied after each x/u/id gate.
Eq.2 : two-qubit depolarizing noise applied after each cx gate.
Eq.3 : crosstalk channel applied ONCE, globally, across all 9 data
       qubits (not per 3-qubit block), injected right after encoding
       and before stabilizer measurement.

       The error operator E belongs to P^9, P = {I, X, Y, Z} -- i.e. E
       is any length-9 string over the alphabet {I,X,Y,Z}, one letter
       per data qubit. E = I^9 (no error at all) is NOT part of this
       channel -- it is given probability 1 - p separately, so p is the
       TOTAL probability that *some* crosstalk error happens.

       Because all 9 physical qubits are used equally (no qubit is
       "special"), the identity of *which* qubits got hit and *which*
       one got which letter doesn't matter for the probability model --
       only the "shape" of the error matters: how many qubits got X,
       how many got Y, how many got Z (the rest are I). Probability
       mass p is split evenly across every distinct non-identity shape,
       and then split evenly again across every specific qubit
       assignment that shares that shape. This means far fewer distinct
       *probability values* need to be reasoned about than the raw
       4**9 - 1 = 262,143 possible error strings would suggest (there
       are only C(9+3,3) - 1 = 219 distinct shapes when all three
       non-identity Paulis are active).

       `pauli_types` lets you pick which non-identity letters are
       allowed to appear in the crosstalk (this is "what crosstalk to
       use"):
           pauli_types="Z"             -> Z-only global crosstalk
           pauli_types=("X", "Z")      -> mixed X/Z crosstalk
           pauli_types=("X", "Y", "Z") -> the fully general E in P^9
       An optional `max_weight` caps how many qubits can be hit at
       once, which is useful to keep the term count tractable when
       using more than one active Pauli letter.
"""

import numpy as np
from itertools import product as iproduct, combinations
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, pauli_error
from qiskit.quantum_info import Statevector, state_fidelity
from qiskit.quantum_info import DensityMatrix
import matplotlib.pyplot as plt
import os

OUTPUT_DIR    = "QEC"
SHOTS         = 5000
FID_SHOTS     = 4
P1            = 0.02
P2            = 0.02
PX            = 0.02
LOGICAL_STATE = "0"   # options: "0", "1"  (Z basis)
                       #          "+", "-"  (X basis)
                       #          "+i","-i" (Y basis)
SEED          = 12345

BLOCKS  = [(0, 1, 2), (3, 4, 5), (6, 7, 8)]
LEADERS = [0, 3, 6]
NUM_DATA_QUBITS = 9

# Single-qubit state prep recipe for qr[0] before it gets spread across
# the 9 physical qubits, the eigenbasis each state belongs to (used to
# pick the correct measurement-basis rotation at decode time), and the
# classical bit that a *correct* round trip should decode to.
STATE_INFO = {
    "0":  {"prep": [],          "basis": "Z", "expect": 0},
    "1":  {"prep": ["x"],       "basis": "Z", "expect": 1},
    "+":  {"prep": ["h"],       "basis": "X", "expect": 0},
    "-":  {"prep": ["x", "h"],  "basis": "X", "expect": 1},
    "+i": {"prep": ["h", "s"],  "basis": "Y", "expect": 0},
    "-i": {"prep": ["h", "sdg"],"basis": "Y", "expect": 1},
}
```

```python
# ══════════════════════════════════════════════════════════════════════
#  CIRCUIT BLOCKS
# ══════════════════════════════════════════════════════════════════════

def build_encoding(logical_state) -> QuantumCircuit:
    """
    logical_state : one of "0", "1"   (Z-eigenbasis, |0>_L / |1>_L)
                            "+", "-"   (X-eigenbasis, |+>_L / |->_L)
                            "+i","-i"  (Y-eigenbasis, |+i>_L / |-i>_L)
    Everything past the state-prep on qr[0] is a fixed linear circuit, so
    preparing qr[0] in any single-qubit state before entangling it produces
    the matching logical version of that state.
    """
    if logical_state not in STATE_INFO:
        raise ValueError(f"Unknown logical_state: {logical_state!r}, "
                          f"expected one of {list(STATE_INFO)}")
    qr = QuantumRegister(9, "data")
    qc = QuantumCircuit(qr, name="encode")
    for gate in STATE_INFO[logical_state]["prep"]:
        getattr(qc, gate)(qr[0])
    qc.cx(qr[0], qr[3])
    qc.cx(qr[0], qr[6])
    qc.h(qr[0]); qc.h(qr[3]); qc.h(qr[6])
    for leader, (a, b, c) in zip(LEADERS, BLOCKS):
        qc.cx(qr[leader], qr[b])
        qc.cx(qr[leader], qr[c])
    for q in qr:
        qc.id(q)
    return qc


def build_stabilizer_measurement() -> QuantumCircuit:
    data = QuantumRegister(9, "data")
    anc  = QuantumRegister(1, "anc")
    syn  = ClassicalRegister(8, "syn")
    qc   = QuantumCircuit(data, anc, syn, name="stabilizer_meas")
    for i, (leader, (a, b, c)) in enumerate(zip(LEADERS, BLOCKS)):
        qc.cx(data[a], anc[0]); qc.cx(data[b], anc[0])
        qc.measure(anc[0], syn[2 * i])
        qc.reset(anc[0])
        qc.cx(data[a], anc[0]); qc.cx(data[c], anc[0])
        qc.measure(anc[0], syn[2 * i + 1])
        qc.reset(anc[0])
    qc.h(anc[0])
    for q in BLOCKS[0] + BLOCKS[1]:
        qc.cx(anc[0], data[q])
    qc.h(anc[0])
    qc.measure(anc[0], syn[6])
    qc.reset(anc[0])
    qc.h(anc[0])
    for q in BLOCKS[1] + BLOCKS[2]:
        qc.cx(anc[0], data[q])
    qc.h(anc[0])
    qc.measure(anc[0], syn[7])
    qc.reset(anc[0])
    return qc


def build_correction() -> QuantumCircuit:
    data = QuantumRegister(9, "data")
    anc  = QuantumRegister(1, "anc")
    syn  = ClassicalRegister(8, "syn")
    qc   = QuantumCircuit(data, anc, syn, name="correct")
    for i, (leader, (a, b, c)) in enumerate(zip(LEADERS, BLOCKS)):
        bit_a, bit_b = 2 * i, 2 * i + 1
        with qc.if_test((syn[bit_a], 1)):
            with qc.if_test((syn[bit_b], 1)):
                qc.x(data[a])
        with qc.if_test((syn[bit_a], 1)):
            with qc.if_test((syn[bit_b], 0)):
                qc.x(data[b])
        with qc.if_test((syn[bit_a], 0)):
            with qc.if_test((syn[bit_b], 1)):
                qc.x(data[c])
    with qc.if_test((syn[6], 1)):
        with qc.if_test((syn[7], 1)):
            qc.z(data[3])
    with qc.if_test((syn[6], 1)):
        with qc.if_test((syn[7], 0)):
            qc.z(data[0])
    with qc.if_test((syn[6], 0)):
        with qc.if_test((syn[7], 1)):
            qc.z(data[6])
    return qc


def build_decoding(logical_state="0") -> QuantumCircuit:
    """
    Inverts the entangling part of build_encoding, then rotates data[0]
    out of whichever eigenbasis (Z/X/Y) `logical_state` belongs to, so the
    final measurement is deterministic for a correct round trip.
    """
    basis = STATE_INFO[logical_state]["basis"]
    data = QuantumRegister(9, "data")
    out  = ClassicalRegister(1, "out")
    qc   = QuantumCircuit(data, out, name="decode")
    for leader, (a, b, c) in zip(LEADERS, BLOCKS):
        qc.cx(data[leader], data[b])
        qc.cx(data[leader], data[c])
    qc.h(data[0]); qc.h(data[3]); qc.h(data[6])
    qc.cx(data[0], data[3])
    qc.cx(data[0], data[6])
    if basis == "X":
        qc.h(data[0])
    elif basis == "Y":
        qc.sdg(data[0]); qc.h(data[0])
    qc.measure(data[0], out[0])
    return qc


def build_full_circuit(logical_state: int, crosstalk_types=None, crosstalk_p: float = None,
                        crosstalk_weights= None,
                        save_rho: bool = False, apply_correction: bool = True) -> QuantumCircuit:
    if crosstalk_p is None:
        crosstalk_p = PX
    data = QuantumRegister(9, "data")
    anc  = QuantumRegister(1, "anc")
    syn  = ClassicalRegister(8, "syn")
    out  = ClassicalRegister(1, "out")
    qc   = QuantumCircuit(data, anc, syn, out)

    qc.compose(build_encoding(logical_state), qubits=data[:], inplace=True)
    qc.barrier(label="Eq.1/Eq.2 noise")

    # ---- SINGLE, GLOBAL crosstalk injection point ----
    # No more qc.append(err.to_instruction(), data[:]) -- that forced Aer to
    # materialize a dense 4**9 x 4**9 superoperator and segfaulted at 7+ qubits.
    # Instead we leave a marker barrier over EXACTLY the 9 data qubits (not the
    # anc qubit, and not the full-width barriers above/below) so a NoiseModel
    # can attach the Pauli-sampled error to it via add_quantum_error.
    if crosstalk_types is not None:
        # FIX: previously this only left a "crosstalk" barrier for a NoiseModel
        # to attach an error to. Two problems made that never fire:
        #   1) Aer's engine treats `barrier` as a pure scheduling no-op and never
        #      routes attached NoiseModel errors through it, at any p.
        #   2) Even swapping the marker to a real instruction (e.g. `delay`)
        #      doesn't fix it: Qiskit silently splits a call like
        #      qc.delay(0, data) into 9 separate 1-qubit `delay` instructions,
        #      so NoiseModel.add_quantum_error's "this exact 9-qubit group" match
        #      never triggers for a correlated multi-qubit Pauli error.
        # Since noise_eq3_crosstalk only ever produces a handful of terms (one
        # representative string per weight, not all 4**9-1 placements), it's
        # cheap to append the QuantumError directly as a genuine 9-qubit
        # instruction on the data qubits -- no dense superoperator, no segfault.
        crosstalk_err = noise_eq3_crosstalk(crosstalk_p, pauli_types=crosstalk_types,
                                             num_qubits=NUM_DATA_QUBITS,
                                             crosstalk_weights=crosstalk_weights)
        qc.append(crosstalk_err.to_instruction(), data[:])
    qc.barrier(label="Eq.3 crosstalk")

    qc.compose(build_stabilizer_measurement(),
               qubits=list(data) + list(anc), clbits=list(syn), inplace=True)
    qc.barrier(label="correction")

    if apply_correction:
        qc.compose(build_correction(),
                   qubits=list(data) + list(anc), clbits=list(syn), inplace=True)

    if save_rho:
        qc.save_density_matrix(qubits=data, label="rho")

    qc.barrier(label="decode")
    qc.compose(build_decoding(logical_state), qubits=data[:], clbits=out[:], inplace=True)
    return qc
```

```python
# ══════════════════════════════════════════════════════════════════════
#  NOISE MODELS
# ══════════════════════════════════════════════════════════════════════

def noise_eq1(p1: float) -> NoiseModel:
    channel = pauli_error([("X", p1/3), ("Y", p1/3), ("Z", p1/3), ("I", 1 - p1)])
    nm = NoiseModel()
    nm.add_all_qubit_quantum_error(channel, ["x", "u", "id"])
    return nm


def noise_eq2(p2: float) -> NoiseModel:
    paulis = ["I", "X", "Y", "Z"]
    terms = [(a + b, p2/15.0) for a, b in iproduct(paulis, paulis) if (a, b) != ("I", "I")]
    terms.append(("II", 1 - p2))
    channel = pauli_error(terms)
    nm = NoiseModel()
    nm.add_all_qubit_quantum_error(channel, ["cx"])
    return nm



def build_eq3_weight_probs(p: float, num_qubits: int = 9,
                            pauli_types=("Z",), crosstalk_weights=None) -> dict:
    """
    One representative string per weight k (qubits are interchangeable,
    order doesn't matter) instead of enumerating all C(num_qubits, k)
    placements. Each requested weight gets an equal share of p, and
    that FULL share goes to its single string (no further division).

    weights : which weights to include, e.g. [1,2,3] or range(1,10).
              Defaults to all weights 1..num_qubits.
    """
    assert 0 <= p <= 1, f"p must be in [0, 1], got {p}"
    if isinstance(pauli_types, str):
        pauli_types = (pauli_types,)
    pauli_types = tuple(pauli_types)

    weights=crosstalk_weights

    if weights is None:
        weights = range(1, num_qubits + 1)
    weights = sorted(set(w for w in weights if 1 <= w <= num_qubits))

    if not weights:
        return {}

    p_each_weight = p / len(weights)

    term_probs = {}
    for k in weights:
        chars = ["I"] * num_qubits
        for i in range(k):
            chars[i] = pauli_types[i % len(pauli_types)]  # cycle letters if >1
        key = "".join(chars)
        term_probs[key] = term_probs.get(key, 0.0) + p_each_weight

    return term_probs


def noise_eq3_crosstalk(p: float, pauli_types=("Z",), num_qubits: int = 9,
                        crosstalk_weights=None):
    """
    Same as before, but now uses ONE canonical string per weight k
    (instead of every C(num_qubits, k) permutation), since duplicate
    bit patterns at different positions are treated as equivalent.

    weights : which weight(s) k to include, e.g. [3] for weight-3 only,
              or [1,2,3,4] for weights 1 through 4. Defaults to all
              weights 1..num_qubits.
    """
    if isinstance(pauli_types, str):
        pauli_types = (pauli_types,)
    for letter in pauli_types:
        assert letter in ("X", "Y", "Z"), "pauli_types entries must be 'X', 'Y', or 'Z'"

    term_probs = build_eq3_weight_probs(p, num_qubits, pauli_types, crosstalk_weights)
    total_p = sum(term_probs.values())
    terms = [(s, prob) for s, prob in term_probs.items()]
    terms.append(("I" * num_qubits, 1.0 - total_p))

    return pauli_error(terms)

def attach_crosstalk(noise_model: NoiseModel = None, crosstalk_types=None, crosstalk_p: float = None,
                      crosstalk_weights= None) -> NoiseModel:
    """
    FIX: crosstalk is no longer attached here via a NoiseModel marker
    instruction -- that mechanism never fired (see build_full_circuit for the
    full explanation). The Eq.3 crosstalk channel is now injected directly
    into the circuit as a real multi-qubit instruction on the data qubits
    (see the `crosstalk_types is not None` branch in build_full_circuit).
    This function is kept so `run()` / `state_fidelity_run()` don't need to
    change: it now just passes the Eq.1/Eq.2 noise_model through untouched.
    crosstalk_types/crosstalk_p/crosstalk_weights are accepted but ignored
    here to preserve the existing call signature.
    """
    return noise_model if noise_model is not None else NoiseModel()
```

```python
# ══════════════════════════════════════════════════════════════════════
#  SIMULATION HELPERS
# ══════════════════════════════════════════════════════════════════════

def run(noise_model, logical_state, shots: int, crosstalk_types=None, crosstalk_p=None,
        crosstalk_weights=None, qc=None, seed: int = SEED, apply_correction: bool = True):
    if qc is None:
        qc = build_full_circuit(logical_state, crosstalk_types=crosstalk_types,
                                 crosstalk_p=crosstalk_p,   # FIX: was never forwarded, so the
                                                             # circuit always used the default PX
                                                             # regardless of the p you asked for
                                 crosstalk_weights=crosstalk_weights,
                                 apply_correction=apply_correction)
    nm = attach_crosstalk(noise_model, crosstalk_types, crosstalk_p, crosstalk_weights)
    sim = AerSimulator(noise_model=nm, seed_simulator=seed)
    return qc, sim.run(qc, shots=shots).result().get_counts()


def state_fidelity_run(logical_state, shots, noise_model=None, crosstalk_types=None, crosstalk_p=None,
                        crosstalk_weights=None, seed=SEED, apply_correction=True) -> float:
    qc = build_full_circuit(logical_state, crosstalk_types=crosstalk_types,
                             crosstalk_p=crosstalk_p,   # FIX: was never forwarded (see run())
                             crosstalk_weights=crosstalk_weights,
                             save_rho=True, apply_correction=apply_correction)
    nm = attach_crosstalk(noise_model, crosstalk_types, crosstalk_p, crosstalk_weights)
    sim = AerSimulator(method="automatic", noise_model=nm, seed_simulator=seed)
    result = sim.run(qc, shots=shots).result()
    rho = result.data(0)["rho"]   # already shot-averaged (pershot=False is the default)
    ideal = Statevector.from_instruction(build_encoding(logical_state))
    return state_fidelity(rho, ideal)


def accuracy(counts: dict, logical_state, shots: int) -> float:
    expected = STATE_INFO[logical_state]["expect"]
    correct = sum(freq for bs, freq in counts.items() if int(bs.split()[0]) == expected)
    return correct / shots * 100


def print_summary(label, acc, fid):
    print(f"  {label:45s}  accuracy={acc:6.2f}%   fidelity={fid:.4f}   infidelity={1-fid:.2e}")


def write_circuit_diagrams(path=None):
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    if path is None:
        path = os.path.join(OUTPUT_DIR, "shor9_circuit_diagrams.txt")
    sections = [
        (f"ENCODING  (logical_state={LOGICAL_STATE!r} -> 9 physical qubits)", build_encoding(LOGICAL_STATE)),
        ("STABILIZER MEASUREMENT  (8 syndrome bits, 1 reused ancilla)", build_stabilizer_measurement()),
        ("CORRECTION  (conditional X/Z fix-up from syndrome)", build_correction()),
        ("DECODING  (9 physical qubits -> 1 logical qubit)", build_decoding()),
        ("FULL PIPELINE  (no crosstalk, Eq.1/Eq.2 noise only, for readability)",
         build_full_circuit(LOGICAL_STATE)),
        ("FULL PIPELINE WITH Eq.3 GLOBAL CROSSTALK  (pX=%.0f%%, Z-only, all 9 qubits)" % (PX * 100),
         build_full_circuit(LOGICAL_STATE, crosstalk_types=("Z",), crosstalk_p=PX)),
    ]
    with open(path, "w", encoding="utf-8") as f:
        f.write("Shor's 9-Qubit Code -- Circuit Diagrams\n")
        f.write("=" * 70 + "\n")
        for title, qc in sections:
            f.write(f"\n\n{'-'*70}\n  {title}\n{'-'*70}\n\n")
            f.write(str(qc.draw(output="text")))
            f.write("\n")
    print(f"Saved circuit diagrams: {path}")
```

```python
if __name__ == "__main__":
    SEP = "=" * 70
    print(SEP)
    print("  Shor's 9-Qubit Code — self-test + noise sweeps")
    print(SEP)

    qc0, counts0 = run(None, LOGICAL_STATE, 512)
    acc0 = accuracy(counts0, LOGICAL_STATE, 512)
    assert acc0 == 100.0, "Noiseless round trip failed -- check encode/correct/decode logic!"

    print(f"\n{SEP}\n  Writing circuit diagrams to file ...\n{SEP}")
    write_circuit_diagrams(path=r"E:\Internship\Internship_3\QEC\QEC\shor9_circuit_diagrams.txt")

    print(f"\n{SEP}\n  Running probability sweeps ...\n{SEP}")

    P_VALS = np.array([0.02, 0.05, 0.058, 0.07, 0.10, 0.20, 0.30, 0.40, 0.55, 0.60, 0.70, 0.80, 0.90])

    # ---- Pick what crosstalk to use here ----
    # "Z"                -> Z-only global crosstalk over all 9 qubits (fast: 511 terms)
    # ("X", "Z")          -> mixed X/Z crosstalk (fast-ish: 3**9 - 1 = 19682 terms)
    # ("X", "Y", "Z")     -> fully general E in {I,X,Y,Z}^9 (slow: 4**9 - 1 = 262143 terms
    #                        -- use CROSSTALK_MAX_WEIGHT below to cap this)
    CROSSTALK_TYPES     = ("Z",)
    CROSSTALK_WEIGHT = [6]   # k

    def sweep_avg_fidelity(kind: str, p_vals, shots=FID_SHOTS):
        fids = []
        for p in p_vals:
            if kind == "eq1":
                f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq1(p))
            elif kind == "eq2":
                f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq2(p))
            elif kind == "eq3":
                f = state_fidelity_run(LOGICAL_STATE, shots,
                                        crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                        crosstalk_weights=CROSSTALK_WEIGHT)
            elif kind == "eq1+eq3":
                f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq1(p),
                                        crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                        crosstalk_weights=CROSSTALK_WEIGHT)
            elif kind == "eq2+eq3":
                f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq2(p),
                                        crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                        crosstalk_weights=CROSSTALK_WEIGHT)
            else:
                raise ValueError(kind)
            fids.append(f)
            print(f"    {kind:10s} p={p:.2f}  ->  fidelity={f:.4f}")
        return np.array(fids)

    curves = {}
    for kind in ("eq1",
                 "eq2",
                 "eq3", "eq1+eq3", "eq2+eq3"):
        print(f"\n  -- sweeping {kind} --")
        curves[kind] = sweep_avg_fidelity(kind, P_VALS)

    fig, ax = plt.subplots(figsize=(9, 7))
    crosstalk_label = "+".join(CROSSTALK_TYPES)
    styles = {
        "eq1":      ("Eq.1 (1-qubit depolarizing)",                              "b-o"),
        "eq2":      ("Eq.2 (2-qubit depolarizing)",                              "g-s"),
        "eq3":      (f"Eq.3 (global {crosstalk_label}-crosstalk, all 9 qubits)", "m-^"),
        "eq1+eq3":  (f"Eq.1 + Eq.3({crosstalk_label}) combined",                 "b--x"),
        "eq2+eq3":  (f"Eq.2 + Eq.3({crosstalk_label}) combined",                 "g--x"),
    }
    for kind, fids in curves.items():
        label, style = styles[kind]
        fid_clipped = np.clip(fids, 1e-6, 1.0)
        ax.loglog(P_VALS, fid_clipped, style, label=label)

    ax.set_xlabel("Error probability p — log scale")
    ax.set_ylabel(f"Average fidelity F(rho, |psi_ideal>) — log scale  ({FID_SHOTS} shots/point)")
    ax.set_title("Shor 9-qubit code: average fidelity vs. probability, per equation")
    ax.legend(fontsize=9)
    ax.grid(True, which="both", alpha=0.3)
    fig.tight_layout()
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    fig.savefig(os.path.join(OUTPUT_DIR, "shor9_fidelity_vs_probability.png"), dpi=150)
    print("\nSaved plot: shor9_fidelity_vs_probability.png")
    print("\nDone.")
```

```python
# ══════════════════════════════════════════════════════════════════════
#  FIDELITY BEFORE CORRECTION
#  (same noise channels as above, but apply_correction=False so the
#  saved density matrix is snapshotted right after the noise/crosstalk
#  and BEFORE the syndrome-based X/Z fix-up is applied)
# ══════════════════════════════════════════════════════════════════════
print(f"\n{SEP}\n  Running probability sweeps WITHOUT correction (pre-correction fidelity) ...\n{SEP}")

def sweep_avg_fidelity_uncorrected(kind: str, p_vals, shots=FID_SHOTS):
    fids = []
    for p in p_vals:
        if kind == "eq1":
            f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq1(p),
                                    apply_correction=False)
        elif kind == "eq2":
            f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq2(p),
                                    apply_correction=False)
        elif kind == "eq3":
            f = state_fidelity_run(LOGICAL_STATE, shots,
                                    crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                    crosstalk_weights=CROSSTALK_WEIGHT,
                                    apply_correction=False)
        elif kind == "eq1+eq3":
            f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq1(p),
                                    crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                    crosstalk_weights=CROSSTALK_WEIGHT,
                                    apply_correction=False)
        elif kind == "eq2+eq3":
            f = state_fidelity_run(LOGICAL_STATE, shots, noise_model=noise_eq2(p),
                                    crosstalk_types=CROSSTALK_TYPES, crosstalk_p=p,
                                    crosstalk_weights=CROSSTALK_WEIGHT,
                                    apply_correction=False)
        else:
            raise ValueError(kind)
        fids.append(f)
        print(f"    {kind:10s} p={p:.2f}  ->  fidelity (before correction)={f:.4f}")
    return np.array(fids)

curves_uncorrected = {}
for kind in ("eq1", "eq2", "eq3", "eq1+eq3", "eq2+eq3"):
    print(f"\n  -- sweeping {kind} (no correction) --")
    curves_uncorrected[kind] = sweep_avg_fidelity_uncorrected(kind, P_VALS)
```

```python
# ══════════════════════════════════════════════════════════════════════
#  ERROR LOCALIZATION FROM THE MEASURED SYNDROME
#  Run the full circuit at a representative error probability, then
#  decode EACH measured 8-bit syndrome using the same classical rules
#  build_correction() uses -- i.e. purely from the syndrome bits, not by
#  peeking at the injected noise -- to work out which physical qubit(s)
#  the stabilizers flagged, and whether it was an X-type or Z-type flag.
#  The result is a histogram over the 9 data qubits: how many shots'
#  syndrome pointed to an error at that qubit.
# ══════════════════════════════════════════════════════════════════════
print(f"\n{SEP}\n  Collecting syndrome measurement statistics ...\n{SEP}")

SYNDROME_P = 0.20   # representative crosstalk probability used for this diagnostic run

qc_syn, counts_syn = run(None, LOGICAL_STATE, SHOTS,
                          crosstalk_types=CROSSTALK_TYPES, crosstalk_p=SYNDROME_P,
                          crosstalk_weights=CROSSTALK_WEIGHT)

# Counts keys come back as "out syn" (space-separated classical registers);
# pull out just the 8-bit syndrome register and tally it on its own.
syndrome_counts = {}
for bitstring, freq in counts_syn.items():
    syn_bits = bitstring.split()[1]
    syndrome_counts[syn_bits] = syndrome_counts.get(syn_bits, 0) + freq

trivial_syndrome = "0" * 8
trivial_frac = syndrome_counts.get(trivial_syndrome, 0) / SHOTS * 100
print(f"  Trivial (all-zero, 'no error detected') syndrome: {trivial_frac:.2f}% of shots")
print(f"  Distinct syndromes observed: {len(syndrome_counts)}")


def decode_syndrome_errors(syn_bits: str):
    """
    Map a measured 8-bit syndrome string to the physical qubit(s) it points
    to, and whether the flag was X-type or Z-type -- using ONLY the syndrome
    bits and the same lookup table build_correction() applies, i.e. as a
    real experiment would decode syndromes without ever seeing the true
    injected error.

    Qiskit prints classical-register bitstrings MSB-first, i.e. syn_bits[0]
    (leftmost char) is syn[7] and syn_bits[-1] (rightmost char) is syn[0],
    so we reverse the string once to index it as bits[i] == syn[i].
    """
    bits = syn_bits[::-1]
    detected = []

    # X-type (bit-flip) checks: one pair of syndrome bits per 3-qubit block
    for i, (leader, (a, b, c)) in enumerate(zip(LEADERS, BLOCKS)):
        bit_a, bit_b = int(bits[2 * i]), int(bits[2 * i + 1])
        if bit_a == 1 and bit_b == 1:
            detected.append((a, "X"))
        elif bit_a == 1 and bit_b == 0:
            detected.append((b, "X"))
        elif bit_a == 0 and bit_b == 1:
            detected.append((c, "X"))
        # (0, 0) -> no X error detected in this block

    # Z-type (phase-flip) check: one representative qubit per block
    s6, s7 = int(bits[6]), int(bits[7])
    if s6 == 1 and s7 == 1:
        detected.append((3, "Z"))
    elif s6 == 1 and s7 == 0:
        detected.append((0, "Z"))
    elif s6 == 0 and s7 == 1:
        detected.append((6, "Z"))
    # (0, 0) -> no Z error detected

    return detected


# Tally, per physical qubit, how many shots' syndrome flagged an X vs. a Z error there.
qubit_error_counts = {q: {"X": 0, "Z": 0} for q in range(NUM_DATA_QUBITS)}
for syn_bits, freq in syndrome_counts.items():
    for qubit, etype in decode_syndrome_errors(syn_bits):
        qubit_error_counts[qubit][etype] += freq

qubits = list(range(NUM_DATA_QUBITS))
x_counts = [qubit_error_counts[q]["X"] for q in qubits]
z_counts = [qubit_error_counts[q]["Z"] for q in qubits]

fig4, ax4 = plt.subplots(figsize=(10, 6))
x_pos = np.arange(NUM_DATA_QUBITS)
width = 0.35
ax4.bar(x_pos - width / 2, x_counts, width, label="X error flagged",
        color="steelblue", edgecolor="black")
ax4.bar(x_pos + width / 2, z_counts, width, label="Z error flagged",
        color="indianred", edgecolor="black")

ax4.set_xticks(x_pos)
ax4.set_xticklabels([f"q{q}" for q in qubits])
ax4.set_xlabel("Physical data qubit")
ax4.set_ylabel("Number of shots the syndrome flagged an error here")
ax4.set_title(f"Errors Localized via Measured Syndrome  "
              f"(p={SYNDROME_P}, {'+'.join(CROSSTALK_TYPES)}-crosstalk, {SHOTS} shots)")
ax4.legend()
ax4.grid(True, axis="y", alpha=0.3)
fig4.tight_layout()
fig4.savefig(os.path.join(OUTPUT_DIR, "syndrome_qubit_localization.png"), dpi=150)
print("Saved plot: syndrome_qubit_localization.png")

print("\n  Per-qubit breakdown (from syndrome decoding only):")
for q in qubits:
    print(f"    q{q}:  X-flagged={x_counts[q]:5d}   Z-flagged={z_counts[q]:5d}")
```

```python
# ══════════════════════════════════════════════════════════════════════
#  INJECTED ERROR ("DISRUPTOR") DISTRIBUTION
#  Visualize the Eq.3 global-crosstalk channel itself: how probability
#  mass is split between "no error" and each error weight k, at the
#  same p used for the syndrome plot above.
# ══════════════════════════════════════════════════════════════════════
print(f"\n{SEP}\n  Visualizing the injected Eq.3 crosstalk error (the 'disruptor') ...\n{SEP}")

weight_probs = build_eq3_weight_probs(SYNDROME_P, NUM_DATA_QUBITS,
                                       pauli_types=CROSSTALK_TYPES,
                                       crosstalk_weights=CROSSTALK_WEIGHT)

weights_used = sorted(set(len(s) - s.count("I") for s in weight_probs))
probs_used = [sum(v for s, v in weight_probs.items() if len(s) - s.count("I") == w)
              for w in weights_used]
identity_prob = 1 - sum(probs_used)

all_labels = ["I (no error)"] + [f"weight {w}" for w in weights_used]
all_probs = [identity_prob] + probs_used

fig5, ax5 = plt.subplots(figsize=(8, 6))
colors = ["gray"] + ["crimson"] * len(weights_used)
ax5.bar(all_labels, all_probs, color=colors, edgecolor="black")
ax5.set_ylabel("Probability")
ax5.set_title(f"Eq.3 Crosstalk Error Distribution  (p={SYNDROME_P}, "
              f"{'+'.join(CROSSTALK_TYPES)}-type, weight(s)={CROSSTALK_WEIGHT})")
for i, v in enumerate(all_probs):
    ax5.text(i, v + 0.01, f"{v:.3f}", ha="center", fontsize=9)
ax5.set_ylim(0, 1.05)
ax5.grid(True, axis="y", alpha=0.3)
fig5.tight_layout()
fig5.savefig(os.path.join(OUTPUT_DIR, "injected_error_distribution.png"), dpi=150)
print("Saved plot: injected_error_distribution.png")

print("\n  Representative error string(s) used in the channel (one per weight):")
for s, p in sorted(weight_probs.items()):
    hit_qubits = [i for i, ch in enumerate(s) if ch != "I"]
    print(f"    weight={len(hit_qubits):>2d}  string={s}  p={p:.4f}  qubits hit={hit_qubits}")
```
